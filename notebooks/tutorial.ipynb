## Setup

First, let's import the necessary modules:

```python
import sys
import os

# Add parent directory to path
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.scenarios.runner import ScenarioRunner
from src.scenarios.comparison import ScenarioComparison
from src.visualization.plots import SimulationVisualizer
from src.reports.report_generator import ReportGenerator

# Set random seed for reproducibility
np.random.seed(42)

print("✓ All modules loaded successfully!")
```

## Part 1: Running a Single Scenario

Let's start by running the baseline scenario (no AI) to understand the basic model.

```python
# Create scenario runner
runner = ScenarioRunner(random_seed=42)

# Run baseline scenario
baseline_results = runner.run_scenario(
    scenario_name='baseline',
    steps=200,
    num_workers=100,
    num_firms=20
)

print("\n" + "="*60)
print("BASELINE SCENARIO COMPLETE")
print("="*60)
```

### Examine Results

Let's look at the time series data:

```python
# View first few rows of results
baseline_df = baseline_results['results']
print("\nTime Series Data (first 10 periods):")
print(baseline_df.head(10))

# View summary statistics
print("\n" + "="*60)
print("SUMMARY STATISTICS")
print("="*60)
for key, value in baseline_results['summary'].items():
    if isinstance(value, dict):
        print(f"\n{key}:")
        for k, v in value.items():
            print(f"  {k}: {v:.4f}")
    else:
        print(f"{key}: {value:.4f}")
```

### Visualize Results

Create plots to understand the dynamics:

```python
# Initialize visualizer
viz = SimulationVisualizer(output_dir='../outputs/figures')

# Create comprehensive dashboard
viz.plot_dashboard(baseline_results)
plt.show()
```

## Part 2: Comparing AI Adoption Scenarios

Now let's compare what happens with different levels of AI adoption.

```python
# Run low adoption scenario
low_results = runner.run_scenario('low_adoption', steps=200)

# Run high adoption scenario
high_results = runner.run_scenario('high_adoption', steps=200)

print("\n✓ All AI adoption scenarios complete!")
```

### Compare Unemployment Across Scenarios

```python
# Create comparison plot
results_dict = {
    'baseline': baseline_results,
    'low_adoption': low_results,
    'high_adoption': high_results
}

viz.plot_comparison(
    results_dict,
    metric='Unemployment Rate',
    title='Unemployment Rate: Impact of AI Adoption'
)
plt.show()
```

### Compare Wage Inequality

```python
# Compare Gini coefficient
viz.plot_comparison(
    results_dict,
    metric='Gini Coefficient',
    title='Income Inequality: Impact of AI Adoption'
)
plt.show()
```

## Part 3: Testing Policy Interventions

Can government policies mitigate the negative effects of AI? Let's find out!

```python
# Run scenario with retraining programs
retraining_results = runner.run_scenario('retraining', steps=200)

# Add to comparison
results_dict['retraining'] = retraining_results

print("\n✓ Policy scenario complete!")
```

### Compare All Scenarios

```python
# Multi-panel comparison
viz.plot_multi_comparison(results_dict)
plt.show()
```

### Quantify Policy Effectiveness

Let's calculate how much retraining programs reduce unemployment:

```python
# Extract steady-state unemployment rates
low_unemployment = low_results['summary']['avg_unemployment_rate']
retraining_unemployment = retraining_results['summary']['avg_unemployment_rate']

reduction = (low_unemployment - retraining_unemployment) * 100

print(f"\nPolicy Impact Analysis:")
print(f"  Low Adoption Unemployment: {low_unemployment*100:.2f}%")
print(f"  With Retraining: {retraining_unemployment*100:.2f}%")
print(f"  \nReduction: {reduction:.2f} percentage points")

# Wage effects
low_wage = low_results['summary']['avg_wage']
retraining_wage = retraining_results['summary']['avg_wage']

wage_increase = ((retraining_wage / low_wage) - 1) * 100

print(f"\nWage Effects:")
print(f"  Low Adoption Wage: ${low_wage:.2f}")
print(f"  With Retraining: ${retraining_wage:.2f}")
print(f"  \nIncrease: {wage_increase:.2f}%")
```

## Part 4: Custom Scenario Design

Design your own scenario with custom parameters!

```python
# Create custom scenario
custom_params = {
    'adoption_rate': 0.6,           # 60% of firms adopt
    'diffusion_speed': 0.05,        # Medium diffusion speed
    'automation_intensity': 0.4,    # Moderate automation
    'augmentation_intensity': 0.5,  # High augmentation
}

custom_results = runner.run_scenario(
    'custom',
    steps=200,
    custom_params=custom_params
)

print("\n✓ Custom scenario complete!")
print("\nResults:")
for key, value in custom_results['summary'].items():
    if not isinstance(value, dict):
        print(f"  {key}: {value:.4f}")
```

## Part 5: Generating Comprehensive Reports

Create a professional report with all scenarios:

```python
# Use ScenarioComparison for full analysis
comparison = ScenarioComparison(random_seed=42)
comparison.results = results_dict

# Create comparison tables
comparison_table = comparison.create_comparison_table()
impact_table = comparison.calculate_impacts()

print("\nScenario Comparison Table:")
print(comparison_table.to_string(index=False))

print("\n" + "="*80)
print("\nImpact Analysis (vs Baseline):")
print(impact_table.to_string(index=False))
```

### Generate HTML Report

```python
# Generate comprehensive report
report_gen = ReportGenerator(output_dir='../outputs/reports')

report_path = report_gen.generate_report(
    results_dict=results_dict,
    comparison_df=comparison_table,
    impact_df=impact_table,
    format='html'
)

print(f"\n✓ Report generated: {report_path}")
print("\nOpen this file in your web browser to view the full report!")
```
